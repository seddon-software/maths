### Trigonometry

```python
import sympy
from IPython.display import display, Math
from sympy import *

d = lambda s: display(Math(s))
x, y, z = sympy.symbols('x, y, z')
O, A, H = sympy.symbols('O, A, H')

def f():
    s = input('>')
    d(f'{s}')
```
